```python
import sqlite3
conn=sqlite3.connect("department.db")
cursor=conn.cursor()
print("Database created sucessfullly")
```

```python
cursor.execute("""
CREATE TABLE Department(
   department_id INTEGER PRIMARY KEY,
   department_name Text,
   location Text
   )
   """)
conn.commit()
print("Table created sucessfully")
```

```python
Values=[
    (1,"HR","Chennai"),
    (2,"IT","Coimbatore"),
    (3,"FINANCE","Sattur"),
    (4,"MARKETING","Sivakasi")
]
cursor.executemany(
    "INSERT INTO Department Values(?,?,?)",
    Values
)
conn.commit()
print("Records inserted sucessfully")
```

```python
cursor.execute("""
CREATE TABLE Employees (
    emp_id INTEGER PRIMARY KEY,
    name TEXT,
    dept_id INTEGER,
    salary INTEGER,
    join_date DATE,
    FOREIGN KEY (dept_id) REFERENCES Departments(dept_id)
)
""")
conn.commit()
print("Table created sucessfully")
```
