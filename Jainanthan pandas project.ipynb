```python
top20 = df.sort_values(by="total_runs",ascending=False).head(20)
print(top20)
```

```python
batsman_runs = df.groupby("batter")["batsman_runs"].sum()
batsman_runs = batsman_runs.sort_values(ascending=False)
print(batsman_runs)
```

```python
season_runs = df.groupby("season")["total_runs"].sum()
print(season_runs)
```

```python
bowler_stats = df.groupby("bowler").agg(
wickets=("wicket", "sum"),
runs_conceded=("total_runs", "sum"),
balls_bowled=("ball", "count")
)
bowler_stats["economy"] = (
    bowler_stats["runs_conceded"] * 6
) / bowler_stats["balls_bowled"]
print(bowler_stats)
```

```python
import matplotlib.pyplot as plt

top10 = df.groupby("batter")["batsman_runs"].sum() \.sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
top10.plot(kind="bar")
plt.title("Top 10 Run Scorers")
plt.xlabel("Batter")
plt.ylabel("Runs")
plt.xticks(rotation=45)
plt.show()
```

```python
top10_wickets = df.groupby("bowler")["wicket"].sum() \.sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
top10_wickets.plot(kind="bar")
plt.title("Top 10 Wicket Takers")
plt.xlabel("Bowler")
plt.ylabel("Wickets")
plt.xticks(rotation=45)
plt.show()
```

```python
dismissal = df["dismissal_kind"].value_counts()
plt.figure(figsize=(8,8))
plt.pie(
    dismissal,
    labels=dismissal.index,
    autopct="%1.1f%%",
    startangle=90,
    shadow=True
)
plt.title("Dismissal Types Distribution")
plt.axis("equal")
plt.show()
```

```python
season_runs = df.groupby("season")["total_runs"].sum()
plt.figure(figsize=(10,5))

plt.plot(
    season_runs.index,
    season_runs.values,
    marker="o",
    linewidth=2
)

plt.title("Season-wise Total Runs")
plt.xlabel("Season")
plt.ylabel("Runs")
plt.grid(True)
plt.show()
```
