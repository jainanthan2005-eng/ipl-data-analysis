```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
categorical_cols = [
    "city",
    "category",
    "seller_id",
    "payment_type",
    "order_status"
]

for col in categorical_cols:
    df[col] = df[col].fillna("Unknown")
```

```python
numeric_cols = ["quantity", "sales"]

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.isnull().sum()
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df.info()
```

```python
df["year"] = df["order_date"].dt.year
```

```python
df["month"] = df["order_date"].dt.month
```

```python
df["order_value"] = df["sales"] * df["quantity"]
```

```python
df.head()
```

```python
df.to_csv("retail_cleaned.csv", index=False)
```

```python
import sqlite3
```

```python
conn = sqlite3.connect("retail.db")
```

```python
df.to_sql("orders", conn, if_exists="replace", index=False)
```

```python
pd.read_sql("SELECT * FROM orders LIMIT 5", conn)
```

```python
pd.read_sql("""
SELECT SUM(order_value) AS total_sales
FROM orders
""", conn)
```

```python
pd.read_sql("""
SELECT COUNT(*) AS total_orders
FROM orders
""", conn)
```

```python
pd.read_sql("""
SELECT category,
       SUM(order_value) AS total_sales
FROM orders
GROUP BY category
ORDER BY total_sales DESC
""", conn)
```

```python
pd.read_sql("""
SELECT city,
       SUM(order_value) AS total_sales
FROM orders
GROUP BY city
ORDER BY total_sales DESC
""", conn)
```

```python
pd.read_sql("""
SELECT seller_id,
       SUM(order_value) AS total_sales
FROM orders
GROUP BY seller_id
ORDER BY total_sales DESC
LIMIT 10
""", conn)
```

```python
pd.read_sql("""
SELECT year,
       month,
       SUM(order_value) AS total_sales
FROM orders
GROUP BY year, month
ORDER BY year, month
""", conn)
```

```python
pd.read_sql("""
SELECT order_status,
       COUNT(*) AS total_orders
FROM orders
GROUP BY order_status
ORDER BY total_orders DESC
""", conn)
```

```python
pd.read_sql("""
SELECT payment_type,
       COUNT(*) AS total_orders,
       SUM(order_value) AS total_sales
FROM orders
GROUP BY payment_type
ORDER BY total_sales DESC
""", conn)
```
